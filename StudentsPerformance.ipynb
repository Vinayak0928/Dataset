```python
import os
import pandas as pd

# 1. Load StudentsPerformance.csv
# Replace 'StudentsPerformance.csv' with your local file path if different
file_path = "StudentsPerformance.csv"
df = pd.read_csv(file_path)

# 2. Display shape, columns, data types, and missing values
print("--- 2. Dataset Overview ---")
print(f"Shape of the dataset: {df.shape}\n")
print("Columns in the dataset:")
print(df.columns.tolist(), "\n")
print("Data types of each column:")
print(df.dtypes, "\n")
print("Missing values in each column:")
print(df.isnull().sum(), "\n")

# 3. Save it as Excel, JSON, and Parquet
# Note: Saving to parquet requires 'pyarrow' or 'fastparquet' installed (pip install pyarrow)
# Note: Saving to excel requires 'openpyxl' installed (pip install openpyxl)
excel_path = "StudentsPerformance.xlsx"
json_path = "StudentsPerformance.json"
parquet_path = "StudentsPerformance.parquet"

df.to_excel(excel_path, index=False)
df.to_json(json_path, orient="records", indent=4)
df.to_parquet(parquet_path, index=False)
print("--- 3. Files saved successfully. ---\n")

# 4. Reload all files
df_excel = pd.read_excel(excel_path)
df_json = pd.read_json(json_path)
df_parquet = pd.read_parquet(parquet_path)
print("--- 4. All files reloaded successfully. ---\n")

# 5. Compare file sizes
print("--- 5. File Size Comparison ---")
files = [file_path, excel_path, json_path, parquet_path]
for f in files:
    size_bytes = os.path.getsize(f)
    size_kb = size_bytes / 1024
    print(f"{f}: {size_kb:.2f} KB")
print()

# 6. Find the average math, reading, and writing scores
print("--- 6. Average Scores ---")
# Standard Kaggle column names use lowercase with spaces
avg_math = df["math score"].mean()
avg_reading = df["reading score"].mean()
avg_writing = df["writing score"].mean()

print(f"Average Math Score: {avg_math:.2f}")
print(f"Average Reading Score: {avg_reading:.2f}")
print(f"Average Writing Score: {avg_writing:.2f}\n")

# 7. Filter students with math score greater than 80
filtered_df = df[df["math score"] > 80]
print("--- 7. Filtering Complete ---")
print(f"Number of students with math score > 80: {len(filtered_df)}\n")

# 8. Save the filtered records as a new CSV file
filtered_csv_path = "Filtered_StudentsPerformance.csv"
filtered_df.to_csv(filtered_csv_path, index=False)
print(
    f"--- 8. Filtered records saved to '{filtered_csv_path}' successfully. ---"
)
```
